```python
fio = input("ФИО Исследователя: ")
date = input("Дата: ")
experiment = input("Эксперимент: ")
output = input("Вывод: ")
title = "Электронный лабораторный журнал"

data1 = f"ФИО исследователя : {fio}"
data2 = f"Дата : {date}"
data3 = f"Эксперимент : {experiment}"
conclusion1 = "Вывод:"
conclusion2 = output

all_lines = [title, data1, data2, data3, conclusion1, conclusion2]
max_len = max(len(line) for line in all_lines)

journal = "+" + "-" * (max_len + 2) + "+\n"
journal += "| " + title.ljust(max_len) + " |\n"
journal += "+" + "-" * (max_len + 2) + "+\n"
journal += "| " + data1.ljust(max_len) + " |\n"
journal += "| " + data2.ljust(max_len) + " |\n"
journal += "| " + data3.ljust(max_len) + " |\n"
journal += "+" + "-" * (max_len + 2) + "+\n"
journal += "| " + conclusion1.ljust(max_len) + " |\n"
journal += "| " + conclusion2.ljust(max_len) + " |\n"
journal += "+" + "-" * (max_len + 2) + "+"

with open("C:\\Users\\Admin\\Downloads\\journal.txt", "w", encoding="utf-8") as file:
    file.write(journal)
```

```python
number_of_capsules = int(input("Общее количество произведенных капсул: "))
packaging = int(input("Количество капсул в одной упаковке: "))
full_packaging =  number_of_capsules // packaging
remains = number_of_capsules % packaging
print("\n--- Отчет фасовочного цеха ---")
print(f"Полных упаковок:\t{full_packaging}")
print(f"Остаток капсул:\t\t{remains}")
```

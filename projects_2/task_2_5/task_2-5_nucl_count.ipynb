```python
dna = input("Последовательность ДНК").upper()
count_A = dna.count("A")
count_T = dna.count("T")
count_C = dna.count("C")
count_G = dna.count("G")
dna_length = len(dna)
nucleotide_content_A = count_A / dna_length * 100
nucleotide_content_T = count_T / dna_length * 100
nucleotide_content_C = count_C / dna_length * 100
nucleotide_content_G = count_G / dna_length * 100
print(f"ДНК: {dna}\nПодсчет каждого нуклеотида:\nA = {count_A}\nT = {count_T}\nC = {count_C}\nG = {count_G}\nДлина ДНК: {dna_length}\nПроцентное содержание A: {nucleotide_content_A:.2f} %\nПроцентное содержание T: {nucleotide_content_T:.2f} %\nПроцентное содержание C: {nucleotide_content_C:.2f} %\nПроцентное содержание G: {nucleotide_content_G:.2f} %")

```

```python
choice = input("Ты витязь на распутье. Куда пойдешь? Прямо, налево или направо?").strip().lower()
if choice == "прямо":
    print("Смерть свою найдешь...")
elif choice == "налево":
    print("Богатым будешь!")
elif choice == "направо":
    print("Женатым будешь!")
else:
    print("Подумай еще. Есть только три дороги")
```
